```python
from langchain_core.tools import tool

@tool
def divide(a:float,b:float)->float:
  """ Divide 2 numbers. Raises Error if Dividing by Zero"""
  if b==0:
    return "Cannot Divide By Zero"
  return a/b
```

```python
result=divide.invoke({"a":2,"b":3})

print(result)
```

```python
print(divide.name)
print(divide.description)
print(divide.args)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
from google.colab import userdata
gemini_key=userdata.get('GEMINI_API_KEY')
```

```python
llm =ChatGoogleGenerativeAI(model="gemini-2.5-flash",api_key=gemini_key)
```

### Tool Binding

```python
llm_with_tools=llm.bind(tools=[divide])
```

### Tool Calling

```python
response=llm_with_tools.invoke("Hi , How are You!")

print(response)
```

```python
response=llm_with_tools.invoke("Divide 10 by 3")

print(response)
```

```python
print(response.tool_calls[0])
```

```python
result=divide.invoke(response.tool_calls[0])

print(result)
```
